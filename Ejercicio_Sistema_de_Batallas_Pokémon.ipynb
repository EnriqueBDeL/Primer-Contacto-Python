**Ejercicio: Sistema de Batallas Pokémon**

El propósito de este ejercicios es desarrollar un sistema simple para simular batallas entre Pokémon. Esto involucra tanto la creación de un diagrama de clases UML como la programación del sistema basado en las especificaciones dadas. Usa conceptos como herencia, sobrecarga, polimorfismo y encapsulamiento.

El sistema define a cada Pokémon por su nombre y tipo (como agua, fuego o planta). Cada Pokémon también tiene una cantidad de salud total, expresada en puntos de salud. Adicionalmente, los Pokémon están equipados con habilidades específicas, que se traducen en movimientos. Estos movimientos son esenciales, ya que contienen información crítica como el nombre del movimiento, su tipo, y el daño que puede causar al oponente.

Durante las batallas, cada Pokémon puede atacar al adversario utilizando alguno de sus movimientos disponibles. El daño infligido por un ataque depende de varios factores, incluyendo la salud del Pokémon oponente y la compatibilidad entre los tipos de los Pokémon involucrados.

Las batallas generalmente se llevan a cabo entre dos Pokémon. El sistema incluye un mecanismo de control que administra el flujo de la batalla y los turnos de los combatientes. Este mecanismo es también responsable de determinar el fin de la batalla, identificando cuándo un Pokémon ha sido derrotado.

```python
import random

class Habilidad:
    def __init__(self, nombreH: str, tipo: str, ataque: int):
        self.__nombreH = nombreH
        self.__tipo = tipo
        self.__ataque = ataque

    def get_nombreH(self):
        return self.__nombreH

    def get_tipo(self):
        return self.__tipo

    def get_ataque(self):
        return self.__ataque

class Pokemon:
    def __init__(self, nombreP: str, tipo: str, salud: int, habilidades: list):
        self.__nombreP = nombreP
        self.__tipo = tipo
        self.__salud = salud
        self.__habilidades = habilidades

    def get_nombreP(self):
        return self.__nombreP

    def get_tipo(self):
        return self.__tipo

    def get_salud(self):
        return self.__salud

    def set_salud(self, valor):
        if valor < 0:
            self.__salud = 0
        else:
            self.__salud = valor

    def get_habilidades(self):
        return self.__habilidades

    def atacar(self, habilidades, salud, tipo, tipoP):
        habilidad = random.choice(habilidades)

        print(f"\n{self.__nombreP} ataca usando {habilidad.get_nombreH()}!")

        multiplicador = 1.0
        if habilidad.get_tipo() == "Agua" and tipo == "Fuego":
            multiplicador = 2.0
        elif habilidad.get_tipo() == "Fuego" and tipo == "Planta":
            multiplicador = 2.0
        elif habilidad.get_tipo() == "Planta" and tipo == "Agua":
            multiplicador = 2.0
        elif habilidad.get_tipo() == "Agua" and tipo == "Planta":
            multiplicador = 0.5
        elif habilidad.get_tipo() == "Fuego" and tipo == "Agua":
            multiplicador = 0.5
        elif habilidad.get_tipo() == "Planta" and tipo == "Fuego":
            multiplicador = 0.5

        dano_final = int(habilidad.get_ataque() * multiplicador)
        nueva_salud = salud - dano_final

        if multiplicador > 1.0:
            print("¡Es super efectivo!")
        elif multiplicador < 1.0:
            print("No es muy efectivo...")

        if nueva_salud < 0:
            return 0
        return nueva_salud


class ControlBatalla:
    def __init__(self, pokemon1: Pokemon, pokemon2: Pokemon):
        self.__pokemon1 = pokemon1
        self.__pokemon2 = pokemon2
        self.__turnos = 0
        self.__fin = False
        self.__derrota1 = False
        self.__derrota2 = False

    def iniciarBatalla(self):
        print("=== INICIO DE LA BATALLA ===")

        atacante = self.__pokemon1
        defensor = self.__pokemon2

        while not self.__fin:
            self.__turnos += 1
            print(f"\n--- Turno {self.__turnos} ---")

            nueva_salud = atacante.atacar(
                atacante.get_habilidades(),
                defensor.get_salud(),
                defensor.get_tipo(),
                atacante.get_tipo()
            )

            defensor.set_salud(nueva_salud)
            print(f"{defensor.get_nombreP()} recibe daño. Salud restante: {defensor.get_salud()}")

            if defensor.get_salud() <= 0:
                if defensor == self.__pokemon1:
                    self.__derrota1 = True
                else:
                    self.__derrota2 = True

                self.finalizarBatalla(True, self.__derrota1, self.__derrota2)
                break

            temp = atacante
            atacante = defensor
            defensor = temp

    def finalizarBatalla(self, fin: bool, derrota1: bool, derrota2: bool):
        self.__fin = fin
        self.__derrota1 = derrota1
        self.__derrota2 = derrota2

        print("\n=== FIN DE LA BATALLA ===")
        print(f"Total de turnos: {self.__turnos}")

        if self.__derrota1:
            print(f"¡{self.__pokemon1.get_nombreP()} se ha debilitado! ¡{self.__pokemon2.get_nombreP()} gana!")
        elif self.__derrota2:
            print(f"¡{self.__pokemon2.get_nombreP()} se ha debilitado! ¡{self.__pokemon1.get_nombreP()} gana!")

if __name__ == "__main__":
    pistola_agua = Habilidad("Pistola Agua", "Agua", 40)
    burbuja = Habilidad("Burbuja", "Agua", 20)
    lanzallamas = Habilidad("Lanzallamas", "Fuego", 50)
    ascuas = Habilidad("Ascuas", "Fuego", 30)

    squirtle = Pokemon("Squirtle", "Agua", 100, [pistola_agua, burbuja])
    charmander = Pokemon("Charmander", "Fuego", 90, [lanzallamas, ascuas])

    batalla = ControlBatalla(squirtle, charmander)
    batalla.iniciarBatalla()
```
