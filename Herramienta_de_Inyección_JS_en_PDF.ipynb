```python
from pypdf import PdfReader, PdfWriter
from google.colab import files
import os

print("=== Herramienta de Inyección JS en PDF ===")

print("\nSube tu PDF original:")
uploaded = files.upload()

if not uploaded:
    print("No subiste ningún archivo. Vuelve a ejecutar la celda.")
else:
    pdf_entrada = list(uploaded.keys())[0]
    print(f"Archivo '{pdf_entrada}' subido correctamente.\n")

    print("--- Configuración del nuevo PDF ---")

    pdf_salida = input(">Introduce el nombre para el nuevo PDF (ej. archivo_final.pdf): ")

    if not pdf_salida.lower().endswith(".pdf"):
        pdf_salida += ".pdf"

    titulo_alerta = input(">Introduce el título de la ventana de alerta: ")

    mensaje_alerta = input(">Escribe el mensaje que quieres que aparezca: ")

    print("\n--- Procesando ---")
    try:
        reader = PdfReader(pdf_entrada)
        writer = PdfWriter()
        writer.append_pages_from_reader(reader)

        js_code = f"""
        app.alert({{
            cMsg: "{mensaje_alerta}",
            cTitle: "{titulo_alerta}",
            nIcon: 1
        }});
        """
        writer.add_js(js_code)

        with open(pdf_salida, "wb") as f:
            writer.write(f)

        print(f"¡Éxito! PDF modificado guardado como '{pdf_salida}'.")

        # print(f"Descargando '{pdf_salida}' a tu ordenador...")
        # files.download(pdf_salida)

    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")
```
