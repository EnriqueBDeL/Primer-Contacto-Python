3. Uno de los problemas del Machine Learning es la interpretabilidad. Los
árboles de decisión permiten interpretar los modelos de una forma muy
fácil. En este apartado, utiliza el clasificador anterior del árbol de decisión
(DecisionTreeClassifier) para visualizar el árbol y ver la importancia de las
variables.



````python
from sklearn import datasets
from sklearn import model_selection
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.tree import export_graphviz
import graphviz
import numpy as np
import pandas as pd



iris = datasets.load_iris() # leemos el dataset


#Creamos los conjuntos de entrenamiento y test.
x = iris.data #Las 4 características
t = iris.target #target 0, 1, 2
tam_test = 0.30 #Tamaño para el test en %
semilla = 7 #semilla para generar aleatoriedad
xe, xt, te, tt = model_selection.train_test_split(x, t, test_size=tam_test,random_state=semilla)
clasificador= DecisionTreeClassifier()

#Predicción
clasificador.fit(xe,te)
obtenido = clasificador.predict(xt)
clasificacion=accuracy_score(tt, obtenido)
print('clasificacion: %.4f' % (clasificacion))
matriz_de_confusion=confusion_matrix(tt,obtenido)
print(matriz_de_confusion)

graphviz.Source(export_graphviz(clasificador, feature_names=iris.feature_names))

importances = pd.DataFrame({'feature':iris.feature_names,'importance':np.round(
clasificador.feature_importances_,4)})
importances = importances.sort_values('importance',ascending=False)
print(importances)
````

4. Utilizando la base de datos breast cancer Wisconsin, haz un estudio de los
cuatro clasificadores anteriores, analizando el % de clasificación, sus
matrices de confusión y visualizando el árbol de decisión. Saca
conclusiones y haz una discusión de los resultados según tu criterio,
eligiendo el mejor clasificador, no sólo ten en cuenta el % de clasificación,
sino la matriz de confusión (ya que no es lo mismo clasificar a alguien como
enfermo y que esté sano que al revés). Fíjate que, para algunos de ellos, cada ver que lo ejecutas obtienes resultados diferentes. Identifica en cuál
de ellos ocurre. Esto es debido a la inicialización aleatoria que presentan.
Para esos clasificadores, deberás realizar 10 simulaciones y dar los
resultados como media y desviación típica.


Nota.- Si alguna de las simulaciones no converge, es decir, no clasifica bien,
por ejemplo que ha obtenido un 30% de clasificación. Esa debe ser
descartada ya que rompe la media

````python
import numpy as np
import pandas as pd
import graphviz
from sklearn import datasets
from sklearn import model_selection
from sklearn import preprocessing
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# --- A. CARGA Y PREPARACIÓN DE DATOS ---
print("--- CARGA DE DATOS ---")
cancer = datasets.load_breast_cancer() # Cargamos la base de datos [cite: 446]
x = cancer.data
t = cancer.target

# Es vital normalizar los datos de entrada para que los clasificadores funcionen bien
x_norm = preprocessing.scale(x)

# Dividimos en entrenamiento y test (30% test)
# Usamos random_state=7 en la división para que los datos de prueba sean siempre los mismos
xe, xt, te, tt = model_selection.train_test_split(x_norm, t, test_size=0.30, random_state=7)

print(f"Total muestras: {len(x)}. Muestras test: {len(xt)}\n")


# --- B. ESTUDIO DE CLASIFICADORES ESTABLES ---
# Estos clasificadores suelen dar el mismo resultado si los datos no cambian
clasificadores = {
    "Support Vector Machine (SVC)": SVC(),
    "Clasificador Bayesiano": GaussianProcessClassifier()
}

for nombre, clf in clasificadores.items():
    clf.fit(xe, te)
    obtenido = clf.predict(xt)
    acc = accuracy_score(tt, obtenido)
    cm = confusion_matrix(tt, obtenido)

    print(f"--- {nombre} ---")
    print('Clasificación: %.4f' % acc)
    print("Matriz de confusión:")
    print(cm)
    print("\n")


# --- C. ÁRBOL DE DECISIÓN (Visualización y resultados) ---
print("--- Árbol de Decisión ---")
# No fijamos random_state para ver si varía, aunque suele ser bastante estable
clf_tree = DecisionTreeClassifier()
clf_tree.fit(xe, te)
obtenido_tree = clf_tree.predict(xt)
print('Clasificación: %.4f' % accuracy_score(tt, obtenido_tree))
print("Matriz de confusión:")
print(confusion_matrix(tt, obtenido_tree))
print("\n* Para ver el árbol gráficamente, ejecuta la celda de abajo con graphviz *")

# Guardamos el dot data para visualizarlo luego
dot_data = export_graphviz(clf_tree, feature_names=cancer.feature_names,
                           class_names=cancer.target_names, filled=True)


# --- D. RED NEURONAL (10 Simulaciones) ---
# El MLPClassifier inicializa sus pesos de forma aleatoria, por lo que cada ejecución cambia[cite: 436, 437].
print("--- Red Neuronal (MLPClassifier) - 10 Simulaciones ---")

resultados_acc = []
matrices_confusion = []
simulaciones_validas = 0
intentos = 0

# Hacemos simulaciones hasta conseguir 10 válidas (descartando las que no convergen) [cite: 438, 439]
while simulaciones_validas < 10 and intentos < 30:
    intentos += 1
    clf_mlp = MLPClassifier(max_iter=500) # Subimos un poco max_iter para ayudar a converger
    clf_mlp.fit(xe, te)
    obtenido_mlp = clf_mlp.predict(xt)
    acc_mlp = accuracy_score(tt, obtenido_mlp)

    # Descartamos si no clasifica bien (ej: por debajo del 80%) [cite: 438, 439]
    if acc_mlp > 0.80:
        resultados_acc.append(acc_mlp)
        matrices_confusion.append(confusion_matrix(tt, obtenido_mlp))
        simulaciones_validas += 1

# Resultados estadísticos
media_acc = np.mean(resultados_acc)
desviacion_acc = np.std(resultados_acc)

print(f"Resultados tras 10 simulaciones válidas:")
print('Clasificación Media: %.4f' % media_acc)
print('Desviación Típica: %.4f' % desviacion_acc)
# Mostramos una de las matrices típicas obtenidas
print("Ejemplo de Matriz de confusión en una simulación:")
print(matrices_confusion[0])
````

5. Ahora trabajaremos sólo con la red neuronal. Vamos a mejorarla ya que hemos
usado los valores de los parámetros por defecto.

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets, model_selection, preprocessing
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# --- 1. PREPARACIÓN DE DATOS (Igual que en el Ejercicio 4) ---
cancer = datasets.load_breast_cancer()
x_norm = preprocessing.scale(cancer.data) # Normalizamos la entrada
t = cancer.target
xe, xt, te, tt = model_selection.train_test_split(x_norm, t, test_size=0.30, random_state=7)


# --- 2. RED NEURONAL POR DEFECTO Y CURVA DE APRENDIZAJE ---
print("--- Red Neuronal (Parámetros Básicos) ---")
# Usamos random_state para que la gráfica te salga igual si repites la celda
clf_basico = MLPClassifier(max_iter=500, random_state=1)
clf_basico.fit(xe, te)

obtenido_basico = clf_basico.predict(xt)
print('Clasificación Básica: %.4f' % accuracy_score(tt, obtenido_basico))

# Dibujamos la curva de aprendizaje como indica la práctica
# Esta curva indica cómo ha ido la red aprendiendo (disminuyendo el error) a través de las épocas.
grafica_basica = pd.DataFrame(clf_basico.loss_curve_).plot(title="Curva de Aprendizaje (Básica)")
plt.show() # Usamos plt.show() para asegurar que se muestre en Colab/Jupyter


# --- 3. RED NEURONAL OPTIMIZADA (Estudio de parámetros) ---
print("\n--- Red Neuronal Optimizada ---")
# Aquí es donde juegas con los parámetros que menciona el PDF
clf_optimizado = MLPClassifier(
    hidden_layer_sizes=(100, 50),  # Dos capas ocultas: una de 100 neuronas y otra de 50
    max_iter=1000,                 # Aumentamos el máximo de iteraciones
    activation='relu',             # Función de activación (puede ser 'logistic', 'tanh', etc.)
    solver='adam',                 # Método de optimización
    learning_rate_init=0.005,      # Tasa de aprendizaje inicial (más alta que el 0.001 por defecto)
    random_state=1
)

clf_optimizado.fit(xe, te)
obtenido_opt = clf_optimizado.predict(xt)
acc_opt = accuracy_score(tt, obtenido_opt)

print('Clasificación Optimizada: %.4f' % acc_opt)
print("Matriz de confusión:")
print(confusion_matrix(tt, obtenido_opt))

# Dibujamos la curva del modelo optimizado para comparar
grafica_opt = pd.DataFrame(clf_optimizado.loss_curve_).plot(title="Curva de Aprendizaje (Optimizada)", color='orange')
plt.show()
````

6. Investiga los argumentos del árbol de decisión, la Support Vector Machine y el
Clasificador bayesiano usado antes. Plantea una arquitectura con sus
parámetros para esos tres algoritmos para el cáncer de mama. Finalmente,
decide cual obtiene mejor precisión (ACC) teniendo también en cuenta la red
neuronal hecha en el ejercicio 5.

````python
from sklearn import datasets, model_selection, preprocessing
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.gaussian_process.kernels import RBF
from sklearn.metrics import accuracy_score, confusion_matrix

# --- PREPARACIÓN DE DATOS ---
cancer = datasets.load_breast_cancer()
x_norm = preprocessing.scale(cancer.data)
t = cancer.target
xe, xt, te, tt = model_selection.train_test_split(x_norm, t, test_size=0.30, random_state=7)

# --- CONFIGURACIÓN DE LAS ARQUITECTURAS OPTIMIZADAS ---

# 1. Árbol de Decisión Optimizado
# Usamos 'entropy' en lugar de 'gini' y limitamos la profundidad (max_depth) para evitar sobreajuste.
clf_tree_opt = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=5,
    min_samples_split=4,
    random_state=7
)

# 2. Support Vector Machine (SVC) Optimizada
# C controla la penalización de errores (valores altos ajustan más).
# probability=True es útil si luego quieres ensamblar modelos.
clf_svc_opt = SVC(
    C=10.0,
    kernel='rbf',
    gamma='scale',
    random_state=7
)

# 3. Clasificador Bayesiano (Gaussian Process) Optimizado
# Usamos un kernel RBF específico y aumentamos n_restarts_optimizer para mejorar el ajuste.
kernel_rbf = 1.0 * RBF(1.0)
clf_gpc_opt = GaussianProcessClassifier(
    kernel=kernel_rbf,
    n_restarts_optimizer=2,
    random_state=7
)

# --- ENTRENAMIENTO Y EVALUACIÓN ---
modelos = {
    "Árbol de Decisión": clf_tree_opt,
    "Support Vector Machine (SVC)": clf_svc_opt,
    "Clasificador Bayesiano": clf_gpc_opt
}

print("--- COMPARATIVA DE MODELOS OPTIMIZADOS ---\n")

for nombre, modelo in modelos.items():
    modelo.fit(xe, te)
    prediccion = modelo.predict(xt)
    acc = accuracy_score(tt, prediccion)
    cm = confusion_matrix(tt, prediccion)

    print(f"** {nombre} **")
    print('Precisión (ACC): %.4f' % acc)
    print("Matriz de confusión:")
    print(cm)
    print("-" * 30)
````

````python
from sklearn import datasets, model_selection, preprocessing
from sklearn.ensemble import VotingClassifier, RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

# --- PREPARACIÓN DE DATOS ---
cancer = datasets.load_breast_cancer()
x_norm = preprocessing.scale(cancer.data)
t = cancer.target
xe, xt, te, tt = model_selection.train_test_split(x_norm, t, test_size=0.30, random_state=7)

# --- 1. DEFINICIÓN DE LOS 3 MEJORES CLASIFICADORES ---
# (Usamos las configuraciones optimizadas de los ejercicios anteriores)
clf1 = SVC(C=10.0, kernel='rbf', gamma='scale', random_state=7)
clf2 = MLPClassifier(hidden_layer_sizes=(100, 50), max_iter=1000, random_state=1)
clf3 = RandomForestClassifier(n_estimators=100, random_state=7)

# --- 2. SISTEMA DE ENSAMBLADO (VOTO POR MAYORÍA) ---
print("--- Sistema de Ensamblado (Voting Classifier) ---")

# voting='hard' significa que toma la decisión por mayoría estricta de votos
ensamblado = VotingClassifier(
    estimators=[('svm', clf1), ('red_neuronal', clf2), ('random_forest', clf3)],
    voting='hard'
)

ensamblado.fit(xe, te)
obtenido_ensamblado = ensamblado.predict(xt)
acc_ensamblado = accuracy_score(tt, obtenido_ensamblado)
cm_ensamblado = confusion_matrix(tt, obtenido_ensamblado)

print('Precisión del Ensamblado (ACC): %.4f' % acc_ensamblado)
print("Matriz de confusión:")
print(cm_ensamblado)
````
